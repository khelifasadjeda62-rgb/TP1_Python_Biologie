```python
heures = [0, 1, 2, 3, 4, 5]
cellules = [100, 150, 210, 300, 430, 600]

print("Heures :", heures)
print("Cellules :", cellules)
```

```python
import matplotlib.pyplot as plt

plt.plot(heures, cellules, marker='o')
plt.xlabel("Heures")
plt.ylabel("Nombre de cellules")
plt.title("Croissance cellulaire")
plt.grid()
plt.show()
```

```python
import numpy as np

cellules_array = np.array(cellules)

print("Moyenne :", np.mean(cellules_array))
print("Maximum :", np.max(cellules_array))
print("Minimum :", np.min(cellules_array))
```

```python
croissance_moyenne = np.mean(np.diff(cellules_array))
print("Croissance moyenne :", croissance_moyenne)
```
